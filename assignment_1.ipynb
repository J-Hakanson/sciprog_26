```c
#include <stdio.h> /* using for printf */

int main(void) { /* main program */
    printf("Question 1: \n\n");
    /* (a) initialising n */
    int n = 150;
    
    /* (d) storing the last 3 numbers in the sequence */
    int n1 = 0;
    int n2 = 0;
    int n3 = 0; 
    
    int char_count = 0; /* tracking the length of each output line */
    int digits = 0; /* tracking number of digits in each n*/
    int j = n;

    /* (b) applying the function repeatedly */
    /* using a while loop that runs until the most recent numbers are 4, 2, 1 as requested */
    while (!(n3 == 4 && n2 == 2 && n1 == 1)) { 

        char_count += printf("%d, ", n); /* printing n */


        /* (c) printing the sequence with a max 40 character per line */
        /* keeping character count per line under 40*/
        
        if (j == 0) {
            digits = 1;
        } 
        else {
            for (j = n; j > 0; j = j / 10) {    /* incorperating number of characters in the newest number into the 40 character per line*/
                digits = j; 
                }
            }

        if (char_count + digits + 2 > 40) { /* starting new line if over 40 characters are in a line already */
            printf("\n");
            char_count = 0;
        }

        /* updating the last 3 numbers */
        n3 = n2;
        n2 = n1;
        n1 = n;
        
        /* Function being run */
        if (n % 2 == 0) { 
            n = n / 2; /* even condition and value */
        } 
        else {
            n = 3 * n + 1; /* odd condition and value*/
        }

    }
    printf("\n");
    return 0;
}
```

```c
#include <stdio.h> /* using for printf */
#include <math.h>  /* using for math */

int main(void){
    int set;    /* counter to run loop 1-3 */
    int n;      /* decalaring variables */
    int i;      /* counter for for loop */

    double a;           /* doubles for these values because we'll be doing math with them */
    double r;
    double S_sum;
    double S_form;

    /* (a) creating a loop that runs from 1-3 */
    for (set = 1; set <= 3; set++) {

        /* (b) designing switch/case construct that selects for different values of n, a, and r */
        switch (set) {

            /* (c) each set of values is here */
            case 1:
                n = 10000;
                a = 2.0;
                r = 0.01;
                break;

            case 2:
                n = 500;
                a = 0.01;
                r = 1.1;
                break;

            case 3:
                n = 100;
                a = 0.0001;
                r = 2.0;
                break;
        }

        S_sum = 0;         /* setting these both equal to zero, might be unnecessary */
        S_form = 0;

        /* (d) finding Sn using three sets of values for summation and formula methods */

        for (i = 0; i <= n; i++) {
            S_sum = S_sum + a * pow(r, i);
        }

        S_form = (a * (1 - pow(r, n + 1))) / (1 - r);

        /* printing results to compare */
        printf("Set %d: \n", set);
        printf("  S_n Summation = %lf \n", S_sum);
        printf("  S_n Formula   = %lf \n", S_form);

    }
    return 0;
}
```

Both methods provide similar results but start differing as the numbers get longer. This is becasue the floating point arithmetic is not exact, specifically in the summation loop. Floating point numbers have a limited precision and the IEEE standard requres operators to return a correctly rounded result. This creates small errors that compound the longer a loop circulates. Becuase the summation loop is running a significant number of times, these errors accumulate giving a slightly different answer than the pure formula.
