```c
#include <stdio.h>
int main(void) {
/* Declare variables */
   int i;
   float sum1, sum2, diff;
   

/* First sum */
   sum1 = 0.0;
   for (i=1; i<=1000; i++) {
      sum1 += 1 / (float)i;
   }


/* Second sum */
   sum2 = 0.0;
   for (i=1000; i>0; i--) {
      sum2 += 1 / (float)i;
   }

   printf(" Sum1=%f\n",sum1);
   printf(" Sum2=%f\n",sum2);

/* Find the difference */
   diff = sum1 - sum2;

   printf(" Difference between the two is %f\n",diff);

   // yes, the two sums are mathematically the same because the order in which real numbers are added together does not change the value of the sum
   // In floating point arithmetic on the computer, they are not equal as you can see there s some difference. This just has to do with rounded results
   // creating small errors that compound and grow with each iteration of a loop

}
```
